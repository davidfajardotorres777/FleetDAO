# 🚚 Demostración e Inspección Interactiva de FleetDAO

Este cuaderno **Jupyter Notebook** permite probar e inspeccionar visualmente cada función de `FleetDAO` sobre MongoDB.

### Funcionalidades clave:
- **Agregar y modificar cualquier variable nueva en vivo**.
- Operaciones **CRUD completas** (`Create`, `Read`, `Update`, `Delete`).
- **Búsqueda dinámica** por atributos personalizados.
- **Resumen analítico ejecutivo** de la flota.


### 1. Conexión al DAO

```python
from dao import FleetDAO
import pandas as pd

# Inicializar conexión al DAO
dao = FleetDAO()
print("✔ Conexión exitosa a FleetDAO!")
```

### 2. Crear Camión y Agregar Variables Personalizadas
Puedes pasar cualquier variable directamente como argumento al método `add_truck` o agregarla después con `add_variable_to_truck` o `update_truck`.

```python
# 1. Crear camión con kwargs directos
truck_id = dao.add_truck(
    brand="Scania R500 V8", 
    capacity_tons=40.0, 
    patente="AB-123-CD", 
    anio=2025
)
print("✔ Camión insertado en MongoDB con ID:", truck_id)

# 2. Agregar variables personalizadas dinámicamente
dao.add_variable_to_truck(truck_id, "custom_gps_tracker", "GPS-9988-X")
dao.update_truck(truck_id, seguro_vencimiento="2027-12-31", chofer_favorito="Carlos Pérez")

# 3. Leer el documento completo guardado en la base de datos
camion = dao.get_truck_by_id(truck_id)
print("\nDocumento guardado en MongoDB:")
for clave, valor in camion.items():
    print(f"  - {clave}: {valor}")
```

### 3. Búsqueda Dinámica por Variables Personalizadas

```python
# Buscar camiones filtrando por el valor de la nueva variable 'chofer_favorito'
resultados = dao.get_trucks_by_variable("chofer_favorito", "Carlos Pérez")
print(f"Camiones encontrados ({len(resultados)}):")
for r in resultados:
    print(f"  * Marca: {r.get('brand')} | Patente: {r.get('patente')} | Chofer: {r.get('chofer_favorito')}")
```

### 4. Visualización en Tabla Pandas DataFrame

```python
# Convertir la lista de camiones directamente a un DataFrame de Pandas
df_camiones = pd.DataFrame(dao.get_trucks())
df_camiones
```

### 5. Resumen Ejecutivo Analítico de la Flota

```python
summary = dao.get_fleet_summary()
summary
```

### 6. Limpieza Final

```python
# Eliminar camión de prueba y cerrar conexión
dao.delete_truck(truck_id)
dao.close()
print("✔ Prueba completada y conexión a MongoDB cerrada.")
```
