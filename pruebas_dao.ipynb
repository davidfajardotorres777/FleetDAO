# 🚚 Demostración e Inspección Interactiva de FleetDAO (en Español)

Este cuaderno **Jupyter Notebook** permite probar e inspeccionar cada método del `FleetDAO` escrito 100% en español sobre MongoDB.

### Métodos Principales:
- `agregar_camion(...)`: Registra un nuevo camión.
- `obtener_camion_por_id(...)`: Lee el documento de MongoDB.
- `agregar_variable_camion(...)`: Agrega variables dinámicas en tiempo real.
- `buscar_camiones_por_variable(...)`: Filtra por cualquier atributo personalizable.
- `eliminar_variable_camion(...)`: Elimina un campo específico con `$unset`.


### 1. Conexión al DAO

```python
from dao import FleetDAO
import pandas as pd

# Conectar a MongoDB
dao = FleetDAO()
print("✔ Conexión exitosa a FleetDAO!")
```

### 2. Crear Camión y Agregar Variables Personalizadas
Puedes pasar cualquier variable directamente como argumento al método `agregar_camion` o agregarlas después con `agregar_variable_camion` o `actualizar_camion`.

```python
# 1. Crear camión
camion_id = dao.agregar_camion(
    brand="Scania R500 V8", 
    capacity_tons=40.0, 
    patente="AB-123-CD", 
    anio=2025
)
print("✔ Camión insertado en MongoDB con ID:", camion_id)

# 2. Agregar variables dinámicas personalizadas
dao.agregar_variable_camion(camion_id, "custom_gps_tracker", "GPS-9988-X")
dao.actualizar_camion(camion_id, seguro_vencimiento="2027-12-31", chofer_favorito="Carlos Pérez")

# 3. Leer el documento completo guardado en MongoDB
camion = dao.obtener_camion_por_id(camion_id)
print("\nDocumento guardado en MongoDB:")
for clave, valor in camion.items():
    print(f"  • {clave}: {valor}")
```

### 3. Búsqueda Dinámica por Variables Personalizadas

```python
# Buscar camiones por el valor de la nueva variable 'chofer_favorito'
resultados = dao.buscar_camiones_por_variable("chofer_favorito", "Carlos Pérez")
print(f"Camiones encontrados ({len(resultados)}):")
for r in resultados:
    print(f"  • Marca: {r.get('brand')} | Patente: {r.get('patente')} | Chofer: {r.get('chofer_favorito')}")
```

### 4. Visualización en Tabla Pandas DataFrame

```python
# Convertir la lista de camiones directamente a un DataFrame de Pandas
df_camiones = pd.DataFrame(dao.obtener_camiones())
df_camiones
```

### 5. Resumen Ejecutivo de la Flota

```python
resumen = dao.obtener_resumen_flota()
resumen
```

### 6. Limpieza Final

```python
# Eliminar camión de prueba y cerrar conexión
dao.eliminar_camion(camion_id)
dao.close()
print("✔ Prueba completada y conexión a MongoDB cerrada.")
```
